### Using the np.where condition

```python
### we want to add a column called "rainfall_group"
### we want to convert the rainfall values to a categorical scale
### if rainfall value is >30mm, "heavy rainfall"
### if rainfall is between 30-20mm "moderate"
### else, "low"

agric_data["Rainfall_group"]=np.where(agric_data["Rainfall_mm"]>30,"Heavy",
                                      np.where(agric_data["Rainfall_mm"]>=20,"moderate","low"))
```
